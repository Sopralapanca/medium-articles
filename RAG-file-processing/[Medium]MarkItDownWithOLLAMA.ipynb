# Importing Libraries

```python
from markitdown import MarkItDown
```

# Processing files with text information

```python
# Istantiate the object
md = MarkItDown()
```

## PPTx files

```python
result = md.convert("/content/Dickinson_Sample_Slides.pptx")
print(result.text_content)
```

## DOCx files

```python
result = md.convert("/content/file-sample_1MB.docx")
print(result.text_content)
```

## HTML files

```python
result = md.convert("/content/sample1.html")
print(result.text_content)
```

## PDF files

It is unlikely that they can be converted to markdown at the moment, they will be converted to plain text.

```python
result = md.convert("/content/somatosensory.pdf")
print(result.text_content)
```

# Processing images

We use an Ollama server to download a vision model for image processing. The server supports an OpenAI-compatible interface to serve the model through the library.

## Setup

```python
!curl https://ollama.ai/install.sh | sh
```

```python
!ollama serve > server.log 2>&1 &
```

Using llava as vision model

```python
%%capture
model = "llava"
!ollama pull {model}
```

```python
%%capture
!pip install openai==1.57.4 ollama==0.4.4
```

## Processing

```python
from openai import OpenAI
client = OpenAI(
    base_url = 'http://localhost:11434/v1',
    api_key='ollama', # required, but unused
)
```

```python
md = MarkItDown(llm_client=client, llm_model=model)
result = md.convert("/content/sampleimage.jpg")
print(result.text_content)
```
